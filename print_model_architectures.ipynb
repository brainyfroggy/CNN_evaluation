```python
import torch
import os
from torchsummary import summary
from torchvision import models

# Path to the directory containing the models
model_dir = 'F:/yujun/projects/CNN_evaluation/model'

# List of model file names
model_files = [
#                r'Amy_IntermediateRoadAttention_lr1e4_stepsize40_bs128_ep50_June92021_epoch3.pth', 
               r'Amy_IntermediateRoadAttention_lr2e4_stepsize80_bs10_ep100_FinetuningIAPS118_June142021_epoch30.pth']
            #    r'best_model_emotion_regression_Amygdala_0521_50epoch_lr4_128bs_7PM_ckvideo_middleframe_train_epoch3_best_R0.3522.pth', 
            #    r'best_model_emotion_regression_VGG_0522_50epoch_lr4_128bs_23M_ckvideo_middleframe_train_normalization_epoch2_best_R0.5468.pth']

# Function to load a checkpoint and print details
def load_checkpoint(filepath):
    checkpoint = torch.load(filepath, map_location=torch.device('cpu'))

    if 'best_R' in checkpoint:
        print("Best R: {:.4f}".format(checkpoint['best_R']))
    if 'epoch' in checkpoint:
        print("Best epoch: {:}".format(checkpoint['epoch']))

    model = checkpoint['model']
    model.load_state_dict(checkpoint['state_dict'])
    for parameter in model.parameters():
        parameter.requires_grad = False

    model.eval()
    return model

# Loop through the model files, load them, and print their architectures
for model_file in model_files:
    model_path = os.path.join(model_dir, model_file)
    print(f'Loading model from {model_path}...')
    model = load_checkpoint(model_path)
    print(f'Architecture of {model_file}:')
    summary(model, input_size=(3, 224, 224))  # Assuming the input size is (3, 224, 224)
    print('\n' + '-'*80 + '\n')
```

```python
import torch
import torchvision.models as models
from torchsummary import summary

# Load the pre-trained VGG16 model
vgg16 = models.vgg16(pretrained=False)

# Print the architecture of the VGG16 model
print("VGG16 Architecture:")
print(vgg16)

# For a more detailed summary, including the number of parameters
summary(vgg16, input_size=(3, 224, 224))  # Assuming the input size is (3, 224, 224)
```
